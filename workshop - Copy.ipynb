# Task 1.
- Calculate Factorial using recursion

```python
def factorial(num):
    if num == 0 or num == 1:
        return 1
    return num * factorial(num -1 )

factorial(5)
```

```python
class FactorialCalculator:

    def calculate(self, num):
        if num == 0 or num == 1:
            return 1
        return num * self.calculate(num - 1)



calc = FactorialCalculator()
print(calc.calculate(5))  
```

# Task 2.
- Write a function that takes a number and return True if this number is prime and False if not.

```python
def isPrime(num):
    if num < 2:
        return False
    for i in range(2, num):
        if num % i == 0:
            return False
    return True

print(isPrime(20))
```

```python
class isprime:
    def __init__(self, num):
        
        self.num = num
    def is_prime(self):
        for i in range(2, self.num):
    
            if self.num % i == 0:
                return False
        return True

ip=isprime(7)
ip.is_prime()
        
        
```

# Task 3.
- Write a function that takes two numbers and return a list of common dividors

```python
def dividors(num1, num2):
    mn = min(num1, num2)
    common_dividors = []
    for i in range(1, mn+1):
        if num1 % i == 0 and num2 % i ==0:
            common_dividors.append(i)
    return common_dividors

dividors(20, 10)
```

```python
def even_fun(num):
    even = []
    for i in range(num):
        if i % 2 == 0:
            even.append(i)
    return even
print(even_fun(10))
```

```python
class MathUtils:

    def get_common_divisors(self, num1, num2):
        mn = min(num1, num2)
        common_divisors = []
        for i in range(1, mn + 1):
            if num1 % i == 0 and num2 % i == 0:
                common_divisors.append(i)
        return common_divisors

    def get_even_numbers(self, limit):
        even_numbers = []
        for i in range(limit):
            if i % 2 == 0:
                even_numbers.append(i)
        return even_numbers



math_tool = MathUtils()
print(
     math_tool.get_common_divisors(20, 10)
)  # Output: [1, 2, 5, 10]
print(
     math_tool.get_even_numbers(10)
)  
```

# Task 4. 
- Given two strings, write a program that efficiently finds the longest common subsequence

- **Ex:** str1 = "Welcome to Machine Learning Diploma" str2 = "I am studying a Machine Learning Course" --> Machine Learning

```python
str1 = "Welcome to Machine Learning Diploma"
str2 = "I am studying a Machine Learning course"

longest_str = ""

for i in range(len(str1)):
    for j in range(i, len(str1)):
        sub_str = str1[i : j + 1]
        if sub_str in str2:
            if len(sub_str) > len(longest_str):
                longest_str = sub_str

print(f"string 1: '{str1}'")
print(f"string 2: '{str2}'")
print("-" * 30)
print(f"The Longest Common Substring is: '{longest_str}'")
```

```python
class StringMatcher:

    def find_longest_common_substring(self, str1, str2):
        longest_str = ""
        for i in range(len(str1)):
            for j in range(i, len(str1)):
                sub_str = str1[i : j + 1]
                if sub_str in str2:
                    if len(sub_str) > len(longest_str):
                        longest_str = sub_str
        return longest_str



matcher = StringMatcher()
s1 = "Welcome to Machine Learning Diploma"
s2 = "I am studying a Machine Learning course"
print(
    matcher.find_longest_common_substring(s1, s2)
)  
```

# Task 5.
- Write a function that will take a given string and reverse the order of words.

- **Ex:** “Hello world” → ”world Hello”


```python
def reverse_words(s):
    words = s.split() 
    reversed_words = words[::-1]  
    return " ".join(reversed_words) 

s = "Hello World"
print(reverse_words(s))
```

```python
class WordReverser:

    def reverse_words(self, text):
        words = text.split()
        reversed_words = words[::-1]
        return " ".join(reversed_words)


reverser = WordReverser()
print(
    reverser.reverse_words("Hello World")
)  
```

# Task 6.
- Create a program that generates a random password of a given length.
- Hint: use random module → import random → random.randrange

```python
import random

class password_gen:
    def __init__(self, length):
        self.length = length
        

    def generate_password(self):
        characters = "abcdefghijklmnopqrstuvwxyzABCDEFGHIJKLMNOPQRSTUVWXYZ1234567890!@#$%^&*"
        num_of_chars = len(characters)
        password = ""
        

        for _ in range(self.length):
            rand_indx = random.randrange(0, num_of_chars)
            password += characters[rand_indx]

        return password
    

if __name__ == "__main__":

    p1 =password_gen(8)
    print(p1.generate_password())
```

```python
# another solution
import random

def generate_password(length):
    characters = "abcdefghijklmnopqrstuvwxyzABCDEFGHIJKLMNOPQRSTUVWXYZ1234567890!@#$%^&*"
    password = ""

    for _ in range(length):
        password += random.choice(characters)

    return password

# Usage example
password_length = 8
random_password = generate_password(password_length)
print(random_password)
```

```python
import random


class PasswordGenerator:

    def __init__(self, length):
        self.length = length
        self.characters = "abcdefghijklmnopqrstuvwxyzABCDEFGHIJKLMNOPQRSTUVWXYZ1234567890!@#$%^&*"

    def generate(self):
        password = ""
        for _ in range(self.length):
            password += random.choice(self.characters)
        return password


pwd_gen = PasswordGenerator(8)
print(pwd_gen.generate())  
```

# Task 7.
- Write a Python program to find all the unique words and count the frequency of occurrence from a given list of strings

- **Ex:** ["Welcome", "Ali", "Hi", "Ali", "No", "Hi", "No", "Ali", "No", "Ali"]  → {'Ali': 4, 'Welcome': 1, 'No': 3, 'Hi': 2}

```python
words = ["Welcome", "Ali", "Hi", "Ali", "No", "Hi", "No", "Ali", "No", "Ali"]
counts = {}
for word in words:
    counts[word] = counts.get(word, 0) + 1
print(counts)
```

```python
# another solution
words = ["Welcome", "Ali", "Hi", "Ali", "No", "Hi", "No", "Ali", "No", "Ali"]
counts = {word: words.count(word) for word in set(words)}

for word in counts:
    print(f"{word}:{counts[word]}")
```

```python
class WordCounter:

    def count_frequencies(self, word_list):
        counts = {}
        for word in word_list:
            counts[word] = counts.get(word, 0) + 1
        return counts



counter = WordCounter()
words_list = [
    "Welcome",
    "Ali",
    "Hi",
    "Ali",
    "No",
    "Hi",
    "No",
    "Ali",
    "No",
    "Ali",
]
print(
    counter.count_frequencies(words_list)
)  
```
